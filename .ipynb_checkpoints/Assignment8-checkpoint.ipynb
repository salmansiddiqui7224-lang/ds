```python

products = {
    "P1": {"name": "Shoes",      "price": 2500, "stock": 5},
    "P2": {"name": "Shirt",      "price": 1500, "stock": 10},
    "P3": {"name": "Bag",        "price": 3000, "stock": 4},
    "P4": {"name": "Watch",      "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}


cart = {}


def view_products():
    print("\n" + "=" * 55)
    print(f"{'ID':<6}{'Product':<15}{'Price':<12}{'Available Stock'}")
    print("=" * 55)
    for pid in products:
        item = products[pid]
        print(f"{pid:<6}{item['name']:<15}Rs. {item['price']:<8}{item['stock']}")
    print("=" * 55)


def add_to_cart():
    view_products()
    pid = input("Enter Product ID (e.g. P1): ").strip().upper()

    if pid not in products:
        print("Error: Product does not exist!")
        return

    qty_text = input("Enter quantity: ").strip()
    if not qty_text.isdigit() or int(qty_text) <= 0:
        print("Error: Please enter a valid quantity (positive number).")
        return
    qty = int(qty_text)

    already_in_cart = cart.get(pid, 0)
    available = products[pid]["stock"] - already_in_cart

    if qty > available:
        print(f"Error: Not enough stock! Only {available} more available.")
        return

    cart[pid] = already_in_cart + qty
    print(f"Added {qty} x {products[pid]['name']} to cart.")


def view_cart():
    if len(cart) == 0:
        print("\nYour cart is empty.")
        return 0

    print("\n" + "=" * 65)
    print(f"{'Product':<15}{'Qty':<8}{'Price':<12}{'Subtotal'}")
    print("=" * 65)
    total = 0
    for pid in cart:
        name = products[pid]["name"]
        price = products[pid]["price"]
        qty = cart[pid]
        subtotal = price * qty
        total += subtotal
        print(f"{name:<15}{qty:<8}Rs. {price:<8}Rs. {subtotal}")
    print("=" * 65)
    print(f"Cart Total: Rs. {total}")
    return total


def checkout():
    if len(cart) == 0:
        print("\nYour cart is empty. Add some items first.")
        return

    total = view_cart()

    discount = 0
    if total >= 5000:
        discount = total * 0.10
        print(f"Discount (10%): - Rs. {discount:.2f}")
    else:
        print("No discount applied (cart total is below Rs. 5000).")

    final_bill = total - discount
    print(f"Final Bill: Rs. {final_bill:.2f}")

    # Reduce stock and empty the cart
    for pid in cart:
        products[pid]["stock"] -= cart[pid]
    cart.clear()

    print("\nCheckout complete! Thank you for shopping with us.")



print("Welcome to the Online Shopping Cart!")

while True:
    print("\n------ MENU ------")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        view_products()
    elif choice == "2":
        add_to_cart()
    elif choice == "3":
        view_cart()
    elif choice == "4":
        checkout()
    elif choice == "5":
        print("Thank you for visiting. Goodbye!")
        break
    else:
        print("Invalid choice! Please enter a number from 1 to 5.")
```
