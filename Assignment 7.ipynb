Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

instance. Then do some updations with it's methods.



```python
class Employee:

    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, years_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.years_of_experience = years_of_experience

    def describe_employee(self):
   
        print("-" * 45)
        print(f"Name               : {self.name}")
        print(f"Employee Number    : {self.number}")
        print(f"Address            : {self.address}")
        print(f"Salary             : {self.salary:,.2f}")
        print(f"Post               : {self.post}")
        print(f"Year of Joining    : {self.year_of_joining}")
        print(f"Education          : {self.education}")
        print(f"Years of Experience: {self.years_of_experience}")
        print("-" * 45)

    def update_year_of_experience(self, years=1):
        if years < 0:
            print("Experience cannot be reduced!")
            return
        self.years_of_experience += years
        print(f"{self.name}'s experience updated to {self.years_of_experience} year(s).")

    def increase_salary(self, percent):
        if percent < 0:
            print("Increment cannot be negative!")
            return
        increment = self.salary * percent / 100
        self.salary += increment
        print(f"{self.name}'s salary increased by {percent}% "
              f"(+{increment:,.2f}). New salary: {self.salary:,.2f}")



employee101 = Employee("Salman Siddiqui", 101, "House 12, Gulshan-e-Iqbal, Karachi",
                       80000, "Software Engineer", 2018, "BS Computer Science", 5)

print("=== employee101 attributes ===")
print("Name  :", employee101.name)
print("Number:", employee101.number)
print("Post  :", employee101.post)
print("Salary:", employee101.salary)

print("\n=== employee101 details ===")
employee101.describe_employee()

print("\n=== employee101 updates ===")
employee101.increase_salary(10)
employee101.update_year_of_experience()
employee101.describe_employee()

employees = [
    Employee("Ahmed Raza", 102, "Gulberg, Lahore", 95000, "Project Manager", 2015, "MBA", 9),
    Employee("Sara Ahmed", 103, "F-10, Islamabad", 70000, "UI Designer", 2020, "BS Design", 3),
    Employee("Usman Tariq", 104, "Saddar, Karachi", 65000, "Accountant", 2019, "B.Com", 4),
    Employee("Fatima Noor", 105, "Model Town, Lahore", 120000, "Data Scientist", 2017, "MS Data Science", 6),
    Employee("Bilal Hussain", 106, "Hayatabad, Peshawar", 55000, "Network Technician", 2021, "DAE Electronics", 2),
    Employee("Ayesha Siddiqui", 107, "Clifton, Karachi", 85000, "HR Manager", 2016, "MBA (HR)", 8),
    Employee("Hamza Malik", 108, "Satellite Town, Rawalpindi", 60000, "QA Tester", 2022, "BS Software Eng.", 1),
    Employee("Zainab Ali", 109, "Johar Town, Lahore", 110000, "Senior Developer", 2014, "MS Computer Science", 10),
    Employee("Imran Shah", 110, "Defence, Karachi", 75000, "Sales Executive", 2018, "BBA", 5),
    Employee("Hina Qureshi", 111, "Bahria Town, Islamabad", 90000, "Marketing Lead", 2016, "MBA (Marketing)", 7),
]

print("\n=== ALL 10 EMPLOYEES ===")
for emp in employees:
    emp.describe_employee()

print("\n=== UPDATES ===")
employees[0].increase_salary(15)
employees[1].increase_salary(8)
employees[2].update_year_of_experience(2)
employees[3].increase_salary(12.5)
employees[3].update_year_of_experience()
employees[4].update_year_of_experience(3)
employees[5].increase_salary(5)
employees[6].increase_salary(20)
employees[7].update_year_of_experience()
employees[8].increase_salary(10)
employees[9].update_year_of_experience(2)

print("\n=== UPDATED DETAILS ===")
for emp in employees:
    emp.describe_employee()
```
