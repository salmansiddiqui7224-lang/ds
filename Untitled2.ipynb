```python
class Student:
    def __init__(self, name, roll_no, math, physics, computer):
        self.name = name
        self.roll_no = roll_no
        # Teeno subjects ke marks ek dictionary mein store kar rahe hain
        self.marks = {
            "Math": math,
            "Physics": physics,
            "Computer": computer
        }

    # 1. Poori detail print karne ke liye
    def describe_student(self):
        total = self.calculate_total()
        percentage = self.calculate_percentage()
        grade = self.get_grade()
        
        print("-" * 35)
        print(f"Student Name : {self.name}")
        print(f"Roll Number  : {self.roll_no}")
        print("Marks        :")
        for subject, score in self.marks.items():
            print(f"   - {subject}: {score}/100")
        print(f"Total Marks  : {total}/300")
        print(f"Percentage   : {percentage:.2f}%")
        print(f"Grade        : {grade}")
        print("-" * 35)

    # 2. Teeno subjects ke marks ka total nikalne ke liye
    def calculate_total(self):
        return sum(self.marks.values())

    # 3. Percentage nikalne ke liye (Assuming har subject 100 marks ka hai)
    def calculate_percentage(self):
        total = self.calculate_total()
        return (total / 300) * 100

    # 4. Percentage ke hisab se grade dene ke liye
    def get_grade(self):
        percentage = self.calculate_percentage()
        if percentage >= 80:
            return "A"
        elif percentage >= 65:
            print( "B")
        elif percentage >= 50:
            return "C"
        else:
            return "Fail"

    # 5. Kisi ek subject ke marks change karne ke liye
    def update_marks(self, subject, new_marks):
        # Subject ka pehla letter capital kar rahe hain taake matching mein masla na ho
        subject = subject.capitalize()
        
        if subject in self.marks:
            if 0 <= new_marks <= 100:
                self.marks[subject] = new_marks
                print(f"Success: {subject} ke marks update kar ke {new_marks} kar diye gaye hain.")
            else:
                print("Error: Marks 0 aur 100 ke darmiyan hone chahiye.")
        else:
            print(f"Error: '{subject}' naam ka koi subject maujood nahi hai.")


# --- Code ko Test karne ke liye Example ---

# Ek student object banayein
student1 = Student("Ali Ahmed", 101, 75, 50, 82)

# 1. Initial details print karein
student1.describe_student()

# 2. Kisi ek subject ke marks update karein
student1.update_marks("Physics", 90)

# 3. Update hone ke baad dobara details dekhein
student1.describe_student()
```
